```python
# 1) Wap to check or Print the following 
# --Palindrome string

n = input("Enter a String:")
if n == n[::-1]:
    print("THE NUMBER IS PALINDROME")
else:
    print("THE NUMBER IS NOT PALINDROME")
```

```python
#----FIBONACCI SERIES
n = int(input("ENTER THE NUMBER OF TERMS:"))
num1 = 0
num2 = 1
print(num1, end = " ")
print(num2, end = " ")
s = 0
for i in range(3, n+1):
    s = num1 + num2
    print(s, end=" ")
    num1 = num2
    num2 = s
```

```python
# ------Factorial
n = int(input("ENTER THE NUMBER:"))
fact = 1
for i in range(1, n+1):
    fact *= i
    print("FACTORIAL:",fact)
```

```python
# Wap to Count the Occurance of characters in a string
n = input("Enter the String:")
dic = {}
for ch in n:
    if ch in dic:
        dic[ch] += 1
    else:
        dic[ch] = 1
        print(dic)
```
