```python
def print1ton(n):
    # Base Case
    if n == 1:
        print(1)
        return

    # Recursive Call
    print1ton(n - 1)

    # Print while returning
    print(n)


n = int(input("Enter n: "))
print1ton(n)
```

```python
# WAP to print n to 1 using RECURSION
def printNto1(n):
    if n == 0:
        return
    print(n)
    printNto1(n - 1)

printNto1(10)
```

```python
# WAP to Print the sum of first N Natural Numbers
def sumn(n):
    if n == 1: #Base case --------
        return 1
    smalloutput = sumn(n-1) # Induction Hypothesis-----------
    return smalloutput + n

sumn(10)
```

```python
# WAP to print the sum of all elements of the recursive array
def array_sum(arr, n):
    # Base case: no elements left
    if n == 0:
        return 0

    # Recursive case
    return arr[n - 1] + array_sum(arr, n - 1)


# Driver code
arr = list(map(int, input("Enter array elements: ").split()))
print("Sum =", array_sum(arr, len(arr)))

```

```python
# Factorial Recursive
def factorial(n):
    if n == 0 :
        return 1
    smalloutput = factorial(n-1)
    return smalloutput * n
n = int(input("Enter a number:"))
factorial(n)
```

```python
# ---Given an array of integers arr[], the task is to find the maximum and minimum elements in the array using the minimum number of comparisons ---
def findarr(arr):
    if len(arr) == 0:
        return
    maxelement = -1
    minelement = 10000
    for i in range (len(arr)):
        if arr[i] > maxelement:
            maxelement = arr[i]
        elif arr[i] < minelement:
            minelement = arr[i]

    return(maxelement,minelement)

arr = [1,7,2,4,10,12,15,4,3]
ans = findarr(arr)
print(ans)

    
```

```python
# ---Kth Smallest: Given an integer array arr[] and an integer k, your task is to find and return the kth smallest element in the given array.-------
def kth_smallest(arr, k): # USING QUICKSELECT TIME COMPLEXITY 0(N)
    def quickselect(left, right):
        pivot = arr[right]
        i = left

        for j in range(left, right):
            if arr[j] <= pivot:
                arr[i], arr[j] = arr[j], arr[i]
                i += 1

        arr[i], arr[right] = arr[right], arr[i]

        if i == k - 1:
            return arr[i]
        elif i > k - 1:
            return quickselect(left, i - 1)
        else:
            return quickselect(i + 1, right)

    return quickselect(0, len(arr) - 1)

# Example
arr = [3, 1, 4, 5, 20, 65]
k = 3
print(kth_smallest(arr, k))


```

```python
#Given an array of length N and an integer x, you need to find if x is present in the array or not. Return true or false.

def findAns(arr, x):
    #base case
    if len(arr) == 0:
        return False

    # induction hypothesis
    ans = findAns(arr[1:], x)
    smallAns = None
    #induction step
    if arr[0] == x:
        smallAns = True
    else:
        smallAns = False
        
    return ans or smallAns

l = [1,2,3,4,5,6,7,8]
k = 2
findAns(l, k)
```

```python
#Check if list is sorted or not (using slicing + RECURSION)

def checkSorted(arr):
    #base case
    if len(arr) <= 1:
        return False

    # Ind Hypo
    
    #induction Step
    if arr[0] > arr[1]:
        return False
    ans = checkSorted(arr[1:])
    return ans

arr = [1,2,3,4,5,6,7,1]
checkSorted(arr)
```

```python
# CHECK if List is SORTED USING SLICING


def checkSorted(arr):
    l = len(arr)             #BASE CASE
    if (l ==0 or l == 1):
        return True
    
    smallerList = arr[1:]    #IH
    
    if arr[0] > arr[1]:      # CHECKING FOR 0th Element
        return False

    
    smallerListOutput = checkSorted(smallerList)      # IH - CALLING RECURSION ON REST OF THE LIST
    if smallerListOutput:
        return True
    else:
        return False
    
l = [1,2,3,4]
print(checkSorted(l))
```

```python
# Return the first index of integer x present in the array (use slicing)
def firstIndex(arr, x):
    #Base Case
    if len(arr) == 0:
        return -1

    #Check for 0th Index

    if arr[0] == x:
        return 0

    smallOutput = firstIndex(arr[1:], x)

    if smallOutput == -1:
        return -1
    else:
        return smallOutput + 1


a = [1,2,3,4,5,6,7,8,9,7]
x = 100
print(firstIndex(a, x))
```

```python
# Return the first index of integer x present in the array (without using slicing)

def firstIndexBetter(arr, x, si):
    #Base Case
    if len(arr) == si:
        return -1

    if arr[si] == x:
        return si

    smallOutput = firstIndexBetter(arr, x, si+1)
    
    return smallOutput

a = [1,2,3,4,5,6,7,8,9,7]
x = 10
print(firstIndexBetter(a, x, 0))
```

```python
# Return the last index of integer x present in the array (use slicing)

def lastIndex(arr, x):
    if len(arr) == 0:
        return -1

    if arr[len(arr)-1] == x:
        return len(arr) - 1

    smallOutput = lastIndex(arr[:len(arr)-1], x)

    if smallOutput == -1:
        return -1
    return smallOutput



a = [1,3,7,8,5,7,7,5,4,3,6,7,5,6]
x = 7
print(lastIndex(a, x))
```

```python
# Return the last index of integer x present in the array (use index reference)
def lastIndexBetter(arr, x, si):
    if si == len(arr):
        return -1

    smallOutput = lastIndexBetter(arr, x, si+1)

    if smallOutput != -1:
        return smallOutput 
    else:
        if arr[si] == x:
            return si
        return -1
        
a = [1,3,7,8,5,7,7,5,4,3,6,7,5,6]
x = 7
print(lastIndexBetter(a, x, 0))
```

```python
# Given a string, compute recursively a new string where all 'x' chars have been removed. xaxbx - ab

def removex(s):
    #base case
    if len(s) == 0:
        return s

    # Induction Hypothesis
    smallOutput = removex(s[1:])

    #induction Step
    if s[0] == 'x':
        return smallOutput
    else:
        return s[0] + smallOutput

removex("xaxbx")
```
