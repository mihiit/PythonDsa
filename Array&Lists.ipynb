```python
# 9) Prime number check
n = int(input("Enter number: "))

if n <= 1:
    print("Not Prime")
else:
    is_prime = True
    for i in range(2, n):
        if n % i == 0:
            is_prime = False
            break

    if is_prime:
        print("Prime Number")
    else:
        print("Not Prime")
```

```python
# 10) HCF --Highest common factor
a = int(input("Enter first number: "))
b = int(input("Enter second number: "))  #------input of both the numbers

hcf = 1                           #-----------Because 1 divides every integer So worst-case GCD is always at least 1.
for i in range(1, min(a, b) + 1): #-----------To find the minimum highest common divisor factor between two numbers
    if a % i == 0 and b % i == 0: #-------------i divides a exactly as well as b exactly
        hcf = i                   #--------------------update the value of hcf

print("HCF:", hcf)                #----------------------------helps in printing the highest common factor
```

```python
# LCM 
a = int(input("Enter first number: "))
b = int(input("Enter second number: ")) #-------------input of both the numbers to calculate the least common multiple

lcm = max(a, b) #------------------------So we start checking from the larger number This avoids useless work

while True:
    if lcm % a == 0 and lcm % b == 0:
        print("LCM:", lcm)
        break
    lcm += 1
```

```python
#finding the equilibrium index
l = [1,2,2,9,3,2]
total_sum = sum(l)
left_sum = 0
for i in range(len(l)):
    right_sum = total_sum - left_sum - l[i]

    if left_sum == right_sum:
        print(i)
        break
        left_sum += l[i]
```

```python
# Equilibrium Index
def equilibrium_index(arr):
    total_sum = sum(arr)
    left_sum = 0

    for i in range(len(arr)):
        total_sum -= arr[i]   # now total_sum = right sum
        if left_sum == total_sum:
            return i
        left_sum += arr[i]

    return -1
```
